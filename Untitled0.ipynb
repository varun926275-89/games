```python

from IPython.display import display, HTML

display(HTML(r"""
<div id="snake-arcade" style="
  background:linear-gradient(135deg,#101827,#172c40);
  padding:20px;border-radius:18px;color:white;
  font-family:Arial;text-align:center;max-width:700px;margin:auto">

  <h1 style="color:#35e88a;margin:5px">🐍 SNAKE ARCADE</h1>
  <p style="color:#a9bfd0">Eat • Grow • Survive</p>

  <div style="display:flex;justify-content:space-around;
  background:#263b50;padding:12px;border-radius:12px;margin-bottom:12px">
    <div>SCORE <b id="score" style="color:#35e88a">0</b></div>
    <div>BEST <b id="best" style="color:#ffd166">0</b></div>
    <div>LEVEL <b id="level" style="color:#5cc8ff">1</b></div>
  </div>

  <div style="position:relative;max-width:600px;margin:auto">
    <canvas id="board" width="600" height="400"
      style="width:100%;display:block;border:2px solid #35e88a;
      border-radius:12px;background:#101b28">
    </canvas>

    <div id="overlay" style="
      position:absolute;inset:0;background:#07111ddd;
      display:flex;flex-direction:column;align-items:center;
      justify-content:center;border-radius:12px">
      <h2 id="message" style="color:#35e88a">SNAKE GAME</h2>
      <p id="submessage">Ready for the challenge?</p>
      <button id="start" style="
        background:#35e88a;color:#102018;border:0;
        padding:13px 35px;border-radius:25px;
        font-weight:bold;font-size:17px;cursor:pointer">
        START GAME
      </button>
    </div>
  </div>

  <p style="color:#a9bfd0">⌨️ Use Arrow Keys or W A S D to move</p>

  <div style="display:grid;grid-template-columns:repeat(3,60px);
  gap:8px;justify-content:center;margin-top:12px">
    <span></span>
    <button class="move" data-dir="up">⬆️</button>
    <span></span>
    <button class="move" data-dir="left">⬅️</button>
    <button class="move" data-dir="down">⬇️</button>
    <button class="move" data-dir="right">➡️</button>
  </div>
</div>

<style>
#snake-arcade .move {
  background:#29465c;color:white;border:1px solid #35e88a;
  border-radius:10px;padding:12px;font-size:20px;cursor:pointer;
}
#snake-arcade button:active {transform:scale(.94)}
</style>

<script>
(function(){
const root=document.getElementById('snake-arcade');
if(root.dataset.ready)return;
root.dataset.ready="1";

const canvas=root.querySelector('#board');
const ctx=canvas.getContext('2d');
const cell=20, cols=30, rows=20;

let snake, food, direction, nextDirection;
let score, best=0, level, running=false, timer;
let particles=[];

const $=id=>root.querySelector('#'+id);

function reset(){
  snake=[{x:8,y:10},{x:7,y:10},{x:6,y:10}];
  direction='right';
  nextDirection='right';
  score=0;
  level=1;
  particles=[];
  createFood();
  updateStats();
}

function createFood(){
  do{
    food={x:Math.floor(Math.random()*cols),
          y:Math.floor(Math.random()*rows)};
  }while(snake.some(p=>p.x===food.x&&p.y===food.y));
}

function updateStats(){
  $('score').textContent=score;
  $('best').textContent=best;
  $('level').textContent=level;
}

function setDirection(d){
  const opposite={up:'down',down:'up',left:'right',right:'left'};
  if(d!==opposite[direction])nextDirection=d;
}

document.addEventListener('keydown',e=>{
  const keys={
    ArrowUp:'up',w:'up',ArrowDown:'down',s:'down',
    ArrowLeft:'left',a:'left',ArrowRight:'right',d:'right'
  };
  if(keys[e.key]){
    e.preventDefault();
    setDirection(keys[e.key]);
  }
});

root.querySelectorAll('.move').forEach(b=>{
  b.addEventListener('click',()=>setDirection(b.dataset.dir));
});

function draw(){
  ctx.fillStyle='#101b28';
  ctx.fillRect(0,0,600,400);

  // Background grid
  ctx.strokeStyle='#1c3040';
  ctx.lineWidth=0.5;
  for(let x=0;x<=600;x+=cell){
    ctx.beginPath();ctx.moveTo(x,0);ctx.lineTo(x,400);ctx.stroke();
  }
  for(let y=0;y<=400;y+=cell){
    ctx.beginPath();ctx.moveTo(0,y);ctx.lineTo(600,y);ctx.stroke();
  }

  // Food
  ctx.shadowColor='#ff5266';
  ctx.shadowBlur=18;
  ctx.fillStyle='#ff5266';
  ctx.beginPath();
  ctx.arc(food.x*cell+10,food.y*cell+10,8,0,Math.PI*2);
  ctx.fill();
  ctx.shadowBlur=0;

  // Snake
  snake.forEach((p,i)=>{
    const gradient=ctx.createLinearGradient(
      p.x*cell,p.y*cell,p.x*cell+cell,p.y*cell+cell
    );
    gradient.addColorStop(0,i===0?'#a2ffbf':'#35e88a');
    gradient.addColorStop(1,'#087c54');

    ctx.fillStyle=gradient;
    ctx.shadowColor='#35e88a';
    ctx.shadowBlur=i===0?13:3;
    ctx.beginPath();
    ctx.roundRect(p.x*cell+1,p.y*cell+1,18,18,5);
    ctx.fill();
    ctx.shadowBlur=0;

    if(i===0){
      ctx.fillStyle='#09251b';
      let eyes;
      if(direction==='right')eyes=[[13,6],[13,14]];
      else if(direction==='left')eyes=[[6,6],[6,14]];
      else if(direction==='up')eyes=[[6,6],[14,6]];
      else eyes=[[6,13],[14,13]];

      eyes.forEach(([x,y])=>{
        ctx.beginPath();
        ctx.arc(p.x*cell+x,p.y*cell+y,2,0,Math.PI*2);
        ctx.fill();
      });
    }
  });
}

function gameOver(){
  running=false;
  clearTimeout(timer);
  $('message').textContent='GAME OVER!';
  $('message').style.color='#ff5266';
  $('submessage').textContent='Your final score: '+score;
  $('start').textContent='PLAY AGAIN';
  $('overlay').style.display='flex';
}

function tick(){
  if(!running)return;

  direction=nextDirection;

  let head={...snake[0]};
  if(direction==='up')head.y--;
  if(direction==='down')head.y++;
  if(direction==='left')head.x--;
  if(direction==='right')head.x++;

  const eating=head.x===food.x&&head.y===food.y;

  if(head.x<0||head.x>=cols||head.y<0||head.y>=rows){
    gameOver();return;
  }

  const body=eating?snake:snake.slice(0,-1);
  if(body.some(p=>p.x===head.x&&p.y===head.y)){
    gameOver();return;
  }

  snake.unshift(head);

  if(eating){
    score+=10;
    best=Math.max(best,score);
    level=1+Math.floor(score/50);
    createFood();
    updateStats();
  }else{
    snake.pop();
  }

  draw();
  timer=setTimeout(tick,Math.max(65,130-(level-1)*10));
}

$('start').addEventListener('click',()=>{
  if(running)return;
  reset();
  $('overlay').style.display='none';
  running=true;
  draw();
  tick();
});

reset();
draw();
})();
</script>
"""))
```
